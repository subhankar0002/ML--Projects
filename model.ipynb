```python
df=df.dropna()
df.isnull().sum()
```

```python
col=["Education","Dt_Customer","Z_CostContact","Z_Revenue","Complain","AcceptedCmp1","AcceptedCmp2","AcceptedCmp3","AcceptedCmp4","AcceptedCmp5","Response",]
df=df.drop(columns=col)
```

```python
df["Age"]=2025-df["Year_Birth"]
df.drop(columns="Year_Birth")
```

```python

df["Family_size"] = df["Kidhome"] + df["Teenhome"]
def add_adults(status):
    if status in ['Absurd', 'Alone', 'Divorced', 'Single', 'Widow', 'YOLO']:
        return 1
    else:
        return 2
df["Family_size"] = df["Family_size"] + df["Marital_Status"].apply(add_adults)
```

```python
df["Total_Spend"] = (
    df["MntWines"] +
    df["MntFruits"] +
    df["MntMeatProducts"] +
    df["MntFishProducts"] +
    df["MntSweetProducts"] +
    df["MntGoldProds"]
)
```

```python
df["Total_Purchases"]=(
    df["NumDealsPurchases"] +
    df["NumWebPurchases"] +
    df["NumCatalogPurchases"] +
    df["NumStorePurchases"] 
)
```

```python
df["Income"] = df["Income"].astype(int)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 8))
df.select_dtypes(include='number').boxplot(rot=90)
# sns.boxplot(y=df["Total_Spend"])
plt.title("Boxplots of all numerical features")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
def remove_outliers(df):
    df_clean = df.copy()
    for col in df_clean.select_dtypes(include='number').columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df_clean = df_clean[(df_clean[col] >= lower) & (df_clean[col] <= upper)]
    return df_clean
df_cleaned= remove_outliers(df)
```

```python
df_cleaned.drop(columns="Year_Birth")

from sklearn.preprocessing import LabelEncoder
df_cleaned['Marital_Status'] = LabelEncoder().fit_transform(df_cleaned['Marital_Status'])
```

```python
from sklearn.preprocessing import StandardScaler
num_cols = df_cleaned.select_dtypes(include='number').columns
scaler = StandardScaler()
df_clean_scaled = pd.DataFrame(scaler.fit_transform(df_cleaned[num_cols]), columns=num_cols)
df_clean_scaled["Marital_Status"] = df_clean_scaled["Marital_Status"].astype(int)
```

```python
important_feature = df_clean_scaled[["Marital_Status", "Income", "Recency", "NumWebVisitsMonth", "Age", "Family_size", "Total_Spend", "Total_Purchases"]]
all_feature=df_clean_scaled.iloc[:, 1:-3]
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
important_feature_pca = pca.fit_transform(important_feature)
all_feature_pca = pca.fit_transform(all_feature)
```

```python
from sklearn.cluster import KMeans
inertia = []
K_range = range(1, 11)
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(important_feature_pca) 
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia (WCSS)')
plt.title('Elbow Method for Optimal K')
plt.grid(True)
plt.show()

important_feature_pca_k=4
```

```python
inertia1 = []
K_range = range(1, 11)
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(all_feature_pca) 
    inertia1.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia (WCSS)')
plt.title('Elbow Method for Optimal K')
plt.grid(True)
plt.show()
all_feature_pca_k=4
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=4, random_state=42) 
kmeans.fit(important_feature_pca)
kmeans.fit(all_feature_pca)
labels_im = kmeans.labels_
labels_all = kmeans.labels_
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
# sns.set(style="whitegrid")
sns.scatterplot(
    x=important_feature_pca[:, 0], y=important_feature_pca[:, 1], hue=labels_im, palette="Set2", s=80, edgecolor='black'
)

# Optional: Plot cluster centers
centroids = kmeans.cluster_centers_
plt.scatter(
    centroids[:, 0], centroids[:, 1],
    s=200, c='red', marker='X', label='Centroids'
)

plt.title("Customer Clusters using KMeans + PCA", fontsize=16)
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score

# For important_features
score_imp = silhouette_score(important_feature_pca, labels_im)

# For all_features
# score_all = silhouette_score(pca_all, kmeans_all.labels_)

print("Silhouette (Important Features):", score_imp)
# print("Silhouette (All Features):", score_all)
```
